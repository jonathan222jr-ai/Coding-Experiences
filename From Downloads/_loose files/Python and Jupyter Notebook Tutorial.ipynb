## Numpy module

We will often use numpy module to matrix and vector computations.

```python
import numpy as np
```

```python
a = np.arange(10)
print(a)
```

We can define a row vector as follows:

```python
a = np.array([[1,1,0,0]])
print(a)
print(a.shape)
```

We can define a column vector as follows:

```python
b = a.T
print(b)
print(b.shape)
```

Similarly, we can define a matrix as follows:

```python
A = np.array([[1,2,3,4],[5,6,7,8]])
print(A)
print(A.shape)
```

We can compute dot product and matrix-vector product as follows:

```python
dotProd = np.dot(a,b)
matVecProd = np.dot(A,b)
print(dotProd)
print(matVecProd)
```

One can extract the entries of the resulting array as follows:

```python
print(dotProd[0][0])
print(matVecProd[0][0])
print(matVecProd[1][0])
```

Zero and identity matrix are defined as follows:

```python
B = np.zeros((3,4))
print(B)
```

```python
idenMat = np.eye(5)
print(idenMat)
print(idenMat[0][0])
print(idenMat[0][1])
```

An array of consecutive values can be defined using linspace:

```python
x = np.linspace(0,1,11)
print(x)
```

Random vectors and matrices are useful for testing.

```python
randVec = np.random.rand(5)
randMat = np.random.rand(5,5)
matVecProd = np.dot(randMat,randVec)
print(randVec)
print(randMat)
```

Sometimes it can be useful to exact diagonal entries of a matrix, or create a diagonal matrix from a vector.

```python
diagPart = np.diag(randMat)
print(diagPart)
```

```python
diagMat = np.diag(randVec)
print(diagMat)
```

Also, it can be useful to extract the upper or lower triangular part of the matrices.

```python
upperMat0 = np.triu(randMat)
upperMat1 = np.triu(randMat,1)
print(upperMat0)
print(upperMat1)
```

```python
lowerMat0 = np.tril(randMat,-1)
lowerMat2 = np.tril(randMat,-2)
print(lowerMat0)
print(lowerMat2)
```

## matplotlib module

This module is very helpful for plotting.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*np.pi,100)
y = np.sin(x)
plt.plot(x,y,'.')
plt.xlabel('x', fontsize='x-large')
plt.ylabel('y', fontsize='x-large')
plt.xlim(0,np.pi)
plt.ylim(0,1)
plt.title('Sine Function')
plt.show()
```

Let's plot two plots together and label them.

```python
plt.plot(x,np.sin(x),'r.',label=r"$y=\sin(x)$")
plt.plot(x,np.cos(x),'b-',label=r"$y=\cos(x)$")
plt.xlabel('x', fontsize='x-large')
plt.ylabel('y', fontsize='x-large')
plt.xlim(0,np.pi)
plt.ylim(0,1)
plt.title('Sine and Cosine Function')
plt.legend()
plt.show()
```
